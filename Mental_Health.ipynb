```python
df['Physical_Activity_Hours'] = df['Physical_Activity_Hours'].clip(lower=0)
```

**skewness**

```python
numeric_col = df.select_dtypes(include='number')
numeric_col.skew()
```

```python
top_contry = df['Country'].value_counts().index[:10].tolist()
```

```python
def group_country(contries):
  if contries in top_contry:
    return contries
  else:
    return 'other'

df['Group_Country'] = df['Country'].apply(group_country)
```

```python
df['Group_Country'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
skewes_col = ['Study_Hours']
other_numeric_col = ['Age','Avg_Daily_Usage_Hours','Daily_Unlocks','Physical_Activity_Hours','Sleep_Hours_Per_Night']
orinal_col = ['Stress_Level']
normal_col = ['Gender','Academic_Level','Most_Used_Platform','Purpose_Of_Use','Group_Country']

feau_col = skewes_col + other_numeric_col + orinal_col + normal_col

x = df[feau_col]
y = df['Mental_Health_Score']

X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
df.columns
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,OneHotEncoder,OrdinalEncoder,FunctionTransformer
from sklearn.compose import ColumnTransformer
skew_pipeline = Pipeline(steps=[
    ("log_transfomer",FunctionTransformer(np.log1p)),
    ("scaler",StandardScaler())
])

plain_pipeline = Pipeline(steps=[
    ("scaler",StandardScaler())
])

ordinal_pipeline = Pipeline(steps=[
    ("ordinal_encoder",OrdinalEncoder(categories=[["Low","Medium","High","Very High"]]))
])

nomial_pipeline = Pipeline(steps=[
    ("one_hot_encoder",OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(transformers=[
    ("skew_transformer",skew_pipeline,skewes_col),
    ("plain_transformer",plain_pipeline,other_numeric_col),
    ("ordinal_transformer",ordinal_pipeline,orinal_col),
    ("nomial_transformer",nomial_pipeline,normal_col)
])
```

```python
df['Stress_Level'].value_counts()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error

lr_pipeline = Pipeline(steps=[
    ("preprocessor",preprocessor),
    ("linear_regression",LinearRegression())
])

lr_pipeline.fit(X_train,y_train)
y_pred = lr_pipeline.predict(X_test)

x_train_pred = lr_pipeline.predict(X_train)

mae = mean_absolute_error(y_test,y_pred)
```

```python
lr_r2_testing = r2_score(y_test,y_pred)
lr_2_training = r2_score(y_train,x_train_pred)
```

```python
print("testing score",lr_r2_testing)
print("training score",lr_2_training)
print("mae",mae)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf_pipeline = Pipeline(steps=[
    ("preprocessor",preprocessor),
    ("random_forest",RandomForestRegressor())
])
rf_pipeline.fit(X_train,y_train)
rf_pred = rf_pipeline.predict(X_test)

rf_train_pred = rf_pipeline.predict(X_train)

rf_r2_testing = r2_score(y_test,rf_pred)
rf_2_training = r2_score(y_train,rf_train_pred)
rf_mae = mean_absolute_error(y_test,rf_pred)
```

```python
print("testing score",rf_r2_testing)
print("training score",rf_2_training)
print("mae",rf_mae)
```

```python
from sklearn.model_selection import RandomizedSearchCV
pram_grid = {
    'random_forest__n_estimators':[100,200,300],
    'random_forest__max_depth' : [5,10,15],
    'random_forest__min_samples_split' : [2,5,10],
    'random_forest__min_samples_leaf' : [1,2,4]
}
random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=pram_grid,
    n_iter=15,
    cv=5,
    scoring='r2',
    random_state=42,
    n_jobs=-1
)
random_search.fit(X_train,y_train)
```

```python
random_search.best_params_
```

```python
rf_best_pipeline = random_search.best_estimator_
rf_best_pred = rf_best_pipeline.predict(X_test)
print(f"r2 score after hyperparameter tuning {r2_score(y_test,rf_best_pred)}")
print(f"mae after hyperparameter tuning {mean_absolute_error(y_test,rf_best_pred)}")
```

```python
from sklearn.metrics import mean_squared_error
lr_rmse = np.sqrt(mean_squared_error(y_test,y_pred))
rf_rmse = np.sqrt(mean_squared_error(y_test,rf_best_pred))

rf_tuned_preds = random_search.best_estimator_.predict(X_test)
rf_tuned_rmse = np.sqrt(mean_squared_error(y_test,rf_tuned_preds))
r2_tuned_trainings_preds = random_search.best_estimator_.predict(X_train)
r2_tuned_trainings = r2_score(y_train,r2_tuned_trainings_preds)
r2_tuned_mae = mean_absolute_error(y_test,rf_tuned_preds)
rf_tuned_r2 = r2_score(y_test,rf_tuned_preds)



results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest', 'Random Forest (Tuned)'],
    'r2_score':[lr_r2_testing,rf_r2_testing,rf_tuned_r2],
    'training_r2_score':[lr_2_training,rf_2_training,r2_tuned_trainings],
    'mae':[mae,rf_mae,r2_tuned_mae],
    'rmse':[lr_rmse,rf_rmse,rf_tuned_rmse]
})
```
